# Laboratorio #3 — Sistemas Inteligentes (INFO1157)

```python
import base64
from pathlib import Path

from IPython.display import HTML, display

OUTPUT = Path("output")


def Show(*files: str, maxWidth: int = 650) -> None:
    width = f"{96 // len(files)}%" if len(files) > 1 else f"min(100%, {maxWidth}px)"
    imgs = [
        f'<img src="data:image/png;base64,{base64.b64encode((OUTPUT / f).read_bytes()).decode()}" '
        f'style="width:{width}; height:auto;">'
        for f in files
    ]
    display(HTML(f'<div style="display:flex; gap:10px; align-items:flex-start;">{"".join(imgs)}</div>'))
```
